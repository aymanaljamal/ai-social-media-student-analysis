This histogram with a Kernel Density Estimation (KDE) curve shows the distribution of `Sleep_Hours` among students, including the central tendency, spread, and overall shape of the data.

### Key Findings & Statistical Trends

* **Central Distribution:** Most students sleep between approximately **5 and 8.5 hours per day**, with the distribution centered around **6.5–6.8 hours**.
* **Approximately Symmetric Distribution:** The data follows a roughly **bell-shaped and balanced distribution**, with no strong skew toward either low or high sleep durations.
* **Range:** Sleep duration ranges from approximately **2 to 11 hours per day**.
* **Extreme Values:** A small number of students report very low sleep duration (**below 4 hours**) or unusually high sleep duration (**above 9.5 hours**).

### Data Insights

* **Typical Sleep Pattern:** Most students have sleep durations around **6–8 hours**, representing the typical range within the dataset.
* **Sleep-Deprived Group:** Students sleeping less than **4–4.5 hours** form a smaller subgroup that may be useful to investigate when studying academic performance and mental health.
* **Balanced Distribution:** Since the variable is approximately symmetric, there is no obvious need for a logarithmic transformation.

### Actionable Takeaways

* **Sleep Monitoring:** Consider examining students with less than **4.5 hours of sleep** as a potentially higher-risk group when analyzing academic and health outcomes.
* **Machine Learning:** `Sleep_Hours` can generally be used directly in many models. For algorithms that benefit from standardized numerical features, `StandardScaler` can be applied.
* **Further Analysis:** Compare `Sleep_Hours` with `Mental_Health_Score`, `Academic_Performance_Score`, and `Academic_Failure_Risk` to investigate possible relationships.


![Distribution of Social Isolation Score](../reports/figures/distribution_Social_Isolation_Score.png)

**Figure Overview**
This histogram overlaid with a Kernel Density Estimation (KDE) curve presents the continuous frequency distribution, central tendency, and symmetric spread for `Social_Isolation_Score` across the surveyed student population.

**Key Findings & Statistical Trends**
* **Symmetric Gaussian Distribution:** The metric exhibits a classic bell-shaped (normal) profile centered around a mean of approximately 4.2, peaking at nearly 680 students per bin.
* **Balanced Spread Across Range:** The scores span symmetrically from a lower bound near 1.0 to an upper tail maximum around 8.0, with the vast majority of students scoring between 2.5 and 6.0.

**Data Insights**
* **Moderate Overall Baseline:** Most students experience moderate levels of social isolation, clustering safely away from the extreme high end of the scale.
* **Symmetric Extremes:** Students feeling virtually no isolation ($\le 2.0$) and those experiencing severe isolation ($\ge 6.5$) form equal, symmetric tail segments, making isolation risk an even-spectrum factor across the cohort.

**Actionable Takeaways**
* **Standard Preprocessing:** Apply standard normalization (`StandardScaler` / Z-score transformation) directly, as the feature adheres to a normal distribution without requiring non-linear transforms.
* **High-Isolation Intervention Threshold:** Set an automated flag for students recording a `Social_Isolation_Score` above 6.0 to trigger wellness check-ins and community integration programs.

![Distribution of Physical Activity Hours](../reports/figures/distribution_Physical_Activity_Hours.png)

**Figure Overview**
This histogram overlaid with a Kernel Density Estimation (KDE) curve illustrates the continuous frequency distribution, zero-inflation spike, and right-tail decay for `Physical_Activity_Hours` across the student cohort.

**Key Findings & Statistical Trends**
* **Zero-Inflation Spike:** A dominant spike appears at exactly $0.0$ hours, capturing over 3,100 students who report zero daily physical activity.
* **Unimodal Secondary Spread & Right Skew:** For active individuals, the distribution forms a broad peak centered between 0.8 and 1.5 hours daily before tapering off in a right-skewed tail up to a maximum of 5.0 hours.

**Data Insights**
* **High Sedentary Proportion:** Over 20% of the surveyed population logs no daily physical activity, presenting a widespread sedentary lifestyle baseline across the cohort.
* **Low Average Engagement:** Even among active students, daily activity remains relatively modest, with very few logging more than 3.0 hours.

**Actionable Takeaways**
* **Binary Feature Engineering:** Create a dedicated indicator variable (`Is_Physically_Active`: 0 vs. 1) to model structural differences between non-active and active students.
* **Distribution Scaling:** Apply a $\log(x + 1)$ transformation or utilize tree-based algorithms to effectively manage the heavy zero mass and positive skew.
