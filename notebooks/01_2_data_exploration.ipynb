* **ارتباط محتمل بالضغط/الإرهاق:** غالباً ما ترتبط الدرجات المنخفضة جداً بقوة بالإجهاد النفسي الحاد والإرهاق غير المُدار أو الغياب المطول.

**Data Insights**
* **Bimodal Risk Division:** Most students demonstrate strong academic competency, but the persistent tail of low-value outliers highlights a vulnerable subgroup experiencing severe academic underperformance.

* **حدود التدخل المبكر الموجهة:** إنشاء مشغل المخاطر الآلي للطلاب الذين يسقطون تحت حد الخط السفلي (~46) لتوفير الدعم الأكاديمي والنفسي الفوري.
* **هندسة الميزات للنمذجة التنبؤية:** تطبيق تقنيات القياس القوي أو التحويل لتخفيف تأثير القيم الشاذة على نماذج الانحدار الخطي.

**Actionable Takeaways**
* **Targeted Early-Intervention Thresholds:** Establish an automated risk trigger for students falling below the lower whisker boundary (~46) to provide immediate academic tutoring and mental health support.
* **Feature Engineering for Predictive Modeling:** Consider applying robust scaling or transformation techniques (e.g., Winsorization or log-based adjustments) to mitigate the influence of extreme low outliers during linear model training.

```python
# إنشاء رسم الصندوق لدرجة الأداء الأكاديمي
# Create box plot for Academic Performance Score
fig, ax = plt.subplots(figsize=(10, 6))

sns.boxplot(data=df, y='Academic_Performance_Score', ax=ax, palette='Set2')
ax.set_title('توزيع درجة الأداء الأكاديمي\nDistribution of Academic Performance Score', fontsize=14, fontweight='bold')
ax.set_ylabel('الدرجة / Score', fontsize=12)
ax.grid(axis='y', alpha=0.3)

# إضافة التعليقات الإحصائية
# Add statistical annotations
q1 = df['Academic_Performance_Score'].quantile(0.25)
q3 = df['Academic_Performance_Score'].quantile(0.75)
median = df['Academic_Performance_Score'].median()
iqr = q3 - q1

stats_text = f'الوسيط / Median: {median:.2f}\\nQ1: {q1:.2f}\\nQ3: {q3:.2f}\\nIQR: {iqr:.2f}'
ax.text(0.98, 0.97, stats_text, transform=ax.transAxes, 
        fontsize=10, verticalalignment='top', horizontalalignment='right',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.savefig('../reports/figures/boxplot_Academic_Performance_Score.png', dpi=300, bbox_inches='tight')
plt.show()

print('إحصائيات درجة الأداء الأكاديمي / Academic Performance Score Statistics:')
print(f'  المتوسط / Mean: {df["Academic_Performance_Score"].mean():.2f}')
print(f'  الوسيط / Median: {median:.2f}')
print(f'  الانحراف المعياري / Std Dev: {df["Academic_Performance_Score"].std():.2f}')
print(f'  الحد الأدنى / Min: {df["Academic_Performance_Score"].min():.2f}')
print(f'  الحد الأقصى / Max: {df["Academic_Performance_Score"].max():.2f}')
```

- **مصفوفة الارتباط / Correlation Matrix:** فحص العلاقات بين ساعات الاستخدام والنوم والأداء الأكاديمي
- **Correlation Matrix:** Examine relationships between usage hours, sleep, and academic performance

- **النمذجة التنبؤية / Predictive Modeling:** بناء نماذج تصنيف المخاطر باستخدام السكان المَعلمة
- **Predictive Modeling:** Build risk classification models using flagged populations

- **تصميم التدخل / Intervention Design:** تطوير برامج موجهة للمجموعات الفرعية المعرضة للمخاطر المحددة
- **Intervention Design:** Develop targeted programs for identified at-risk subgroups

- **التتبع الطولي / Longitudinal Tracking:** تطبيق أنظمة المراقبة للمؤشرات الحمراء المبكرة
- **Longitudinal Tracking:** Implement monitoring systems for early warning indicators
